```python
import pandas as pd
import numpy as np

url = 'https://raw.githubusercontent.com/katiegaertner/ML-Fundamentals/main/world-health-data-numbers.csv'
public_health_info = pd.read_csv(url)
```

```python
# rows & columns in the data frame
public_health_info.shape
```

```python
#update notebook
public_health_info.head()
```

```python
# set abbreviation as the index
public_health_info.set_index('Abbreviation', inplace=True)
```

```python
# sort data frame by abbreviation
public_health_info.sort_index(inplace=True)
```

```python
# DZ index value
public_health_info.loc['DZ']
```

```python
# delete the vatican city abbreviation
public_health_info = public_health_info.drop('VA')
```

```python
# language of the nation on the 186 row
public_health_info.iloc[185]
```

```python
# continent with the highest infant mortality
public_health_info.sort_values('Infant mortality_per_1k', ascending=False).head(10)
```

```python
# spanish speaking nations
spanish_speaking_nations = public_health_info[public_health_info['Official Language'] == 'Spanish']

spanish_speaking_nations.shape
```

```python
# mean infant mortality rate spanish speaing nations
spanish_speaking_nations['Infant mortality_per_1k'].mean()
```

```python
# median maternal mortality rate spanish speaking nations
spanish_speaking_nations['Maternal_Mortality_per_100k'].median()
```

```python
# column for GDP
public_health_info['GDP_per_capita'] = (public_health_info['GDP'] / public_health_info['Population'])

public_health_info.sort_values('GDP_per_capita', ascending=False).head(10)
```

```python
# maternial mortality rate per 100,000 live births - switzerland
public_health_info.loc['CH', 'Maternal_Mortality_per_100k']
```

```python
# average maternial mortality rate per 100,000 live births
public_health_info['Maternal_Mortality_per_100k'].mean()
```

```python
# percentage of health care is paid out of pocket - lowest
public_health_info.sort_values('Infant mortality_per_1k', ascending=True).head(1)
```
